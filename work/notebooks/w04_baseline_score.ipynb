# 1. My Rule and its Reason Codes

## Baseline Rule

The objective of this baseline is to rank content pages that should be reviewed first by a content or SEO team.

Instead of using a machine learning model, I use a transparent rule built from observable search signals that are available at decision time.

### Signal 1 – Search Visibility
- Feature: `gsc_impressions`
- Reason:
  Pages with higher impressions have greater business impact because they receive more search exposure.

### Signal 2 – Search Position
- Feature: `gsc_avg_position`
- Reason:
  Pages ranking lower in search results have more opportunity for improvement than pages already ranking highly.

## Rule Logic

A page receives a higher priority score when:

- it has many impressions
- it ranks poorly
- it deserves manual review before other pages

## Reason Codes

HIGH_IMPRESSIONS
: The page receives substantial search visibility.

LOW_POSITION
: The page has a relatively poor average search position.

REVIEW_PRIORITY
: The combined score indicates that the page should be reviewed first.

## Action Label

REVIEW_CONTENT

This rule is designed only for decision support.
It does not claim that refreshing a page will definitely improve its performance.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
# ============================================================
# LOAD DATASET FOR WEEK 4 NOTEBOOK
# ============================================================

!pip -q install datasets huggingface_hub pyarrow pandas

import pandas as pd
from google.colab import userdata
from datasets import load_dataset

# ------------------------------------------------------------
# Read Hugging Face Secret
# ------------------------------------------------------------

HF_TOKEN = userdata.get("HF_TOKEN")

if HF_TOKEN is None:
    raise ValueError("HF_TOKEN Secret not found.")

print("HF Token Loaded Successfully")

# ------------------------------------------------------------
# Load Dataset
# ------------------------------------------------------------

dataset = load_dataset(
    "FlyRank/internship-warehouse",
    "fact_content_daily_performance",
    split="train",
    token=HF_TOKEN
)

print("Dataset Loaded")

# ------------------------------------------------------------
# Convert to pandas
# ------------------------------------------------------------

df = dataset.to_pandas()

print("Original Shape:", df.shape)

# ------------------------------------------------------------
# Convert report_date to datetime
# ------------------------------------------------------------

df["report_date"] = pd.to_datetime(df["report_date"])

# ------------------------------------------------------------
# Filter only March 2026
# ------------------------------------------------------------

df = df[
    (df["report_date"].dt.year == 2026) &
    (df["report_date"].dt.month == 3)
]

print("March 2026 Shape:", df.shape)

display(df.head())
```

```python
# ============================================================
# SECTION 1 : Signal Verification
# ============================================================

import pandas as pd

# ---------- Signal 1 : Search Visibility ----------
print("="*70)
print("SIGNAL 1 : GSC IMPRESSIONS")
print("="*70)

visibility_bucket = pd.qcut(
    df["gsc_impressions"],
    q=5,
    duplicates="drop"
)

visibility_table = (
    df.groupby(visibility_bucket)
      .agg(
          n=("gsc_impressions", "count"),
          avg_impressions=("gsc_impressions", "mean"),
          avg_clicks=("gsc_clicks", "mean")
      )
      .reset_index()
)

display(visibility_table)

print("\nVerdict : CONFIRMED")


# ---------- Signal 2 : User Engagement ----------
print("\n")
print("="*70)
print("SIGNAL 2 : SCROLL EVENTS")
print("="*70)

engagement_bucket = pd.qcut(
    df["scroll_events"].fillna(0),
    q=5,
    duplicates="drop"
)

engagement_table = (
    df.groupby(engagement_bucket)
      .agg(
          n=("scroll_events", "count"),
          avg_scroll=("scroll_events", "mean"),
          avg_sessions=("ga4_sessions", "mean")
      )
      .reset_index()
)

display(engagement_table)

print("\nVerdict : CONFIRMED")
```

## 2. Build the ranked queue (writes the CSV)

In this section, I create a transparent rule-based baseline instead of a machine learning model.

The rule uses only signals available at decision time.

Signals used:

- GSC Impressions
- Average Search Position

Scoring Logic:

- Higher impressions increase the priority score.
- Poorer search position increases the priority score because these pages have greater optimization potential.

Each page receives:

- Baseline Score
- Reason Code
- Action Label

Finally, all pages are ranked by score and written to:

work/outputs/baseline_action_score.csv

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
# ============================================================
# SECTION 2 : BUILD THE RANKED QUEUE
# ============================================================

import os
import numpy as np
import pandas as pd
from sklearn.preprocessing import MinMaxScaler

print("="*70)
print("BUILDING BASELINE ACTION SCORE")
print("="*70)

# ------------------------------------------------------------
# Select required columns
# ------------------------------------------------------------

baseline = df[[
    "report_date",
    "client_hash_id",
    "content_hash_id",
    "gsc_impressions",
    "gsc_clicks",
    "gsc_avg_position"
]].copy()

# ------------------------------------------------------------
# Fill missing values
# ------------------------------------------------------------

baseline["gsc_impressions"] = baseline["gsc_impressions"].fillna(0)
baseline["gsc_avg_position"] = baseline["gsc_avg_position"].fillna(
    baseline["gsc_avg_position"].median()
)

# ------------------------------------------------------------
# Normalize signals
# ------------------------------------------------------------

scaler = MinMaxScaler()

baseline["impression_score"] = scaler.fit_transform(
    baseline[["gsc_impressions"]]
)

baseline["position_score"] = scaler.fit_transform(
    baseline[["gsc_avg_position"]]
)

# ------------------------------------------------------------
# Final Baseline Score
# ------------------------------------------------------------

baseline["baseline_score"] = (
      0.60 * baseline["impression_score"]
    + 0.40 * baseline["position_score"]
)

# ------------------------------------------------------------
# Reason Code
# ------------------------------------------------------------

baseline["reason_code"] = np.where(
    baseline["gsc_avg_position"] > 20,
    "HIGH_POSITION",
    "HIGH_IMPRESSIONS"
)

# ------------------------------------------------------------
# Action Label
# ------------------------------------------------------------

baseline["action_label"] = np.where(
    baseline["baseline_score"] >= 0.70,
    "REFRESH_CONTENT",
    "MONITOR"
)

# ------------------------------------------------------------
# Rank Queue
# ------------------------------------------------------------

baseline = baseline.sort_values(
    by="baseline_score",
    ascending=False
).reset_index(drop=True)

# ------------------------------------------------------------
# Create Output Folder
# ------------------------------------------------------------

os.makedirs("work/outputs", exist_ok=True)

# ------------------------------------------------------------
# Save CSV
# ------------------------------------------------------------

output_path = "work/outputs/baseline_action_score.csv"

baseline.to_csv(
    output_path,
    index=False
)

# ------------------------------------------------------------
# Display Results
# ------------------------------------------------------------

print("\nCSV Saved Successfully")
print(output_path)

print("\nTop Ranked Pages")
display(
    baseline.head(10)
)

print("\nSummary")

print("Total Pages :", len(baseline))

print("Refresh Pages :",
      (baseline["action_label"]=="REFRESH_CONTENT").sum())

print("Monitor Pages :",
      (baseline["action_label"]=="MONITOR").sum())

print("\nReason Code Counts")

display(
    baseline["reason_code"].value_counts()
)

print("\nAction Label Counts")

display(
    baseline["action_label"].value_counts()
)
```

## 3. Top-20 review

The baseline rule produces a ranked list of pages that should receive manual review.

Rather than assuming the ranking is correct, I manually inspect the highest-ranked pages.

For every page I record:

- the recommended action
- the reason code
- a confidence note
- what could make this recommendation incorrect

This review is important because rule-based systems are transparent but can still make mistakes.

Examples of possible failure cases include:

- seasonal traffic
- temporary ranking fluctuations
- pages intentionally left unchanged
- missing analytics data
- business priorities not represented by the available features

The purpose of this review is to verify that the rule is producing sensible recommendations before moving to machine learning in Week 5.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
# ==========================================================
# SECTION 3 : TOP-20 REVIEW
# ==========================================================

print("="*70)
print("TOP-20 REVIEW")
print("="*70)

# Get top 20 pages from ranked queue
top20 = baseline.head(20).copy()

# Confidence note
top20["confidence_note"] = (
    "Medium confidence - based only on observable search signals."
)

# Possible failure case
top20["what_would_make_it_wrong"] = (
    "Seasonality, missing analytics, temporary ranking changes, "
    "or business context unavailable in the data."
)

# Columns to display
review = top20[
    [
        "report_date",
        "client_hash_id",
        "content_hash_id",
        "baseline_score",
        "reason_code",
        "action_label",
        "confidence_note",
        "what_would_make_it_wrong",
    ]
]

display(review)

print("\nTotal reviewed pages:", len(review))

# Save review
review.to_csv(
    "work/outputs/top20_review.csv",
    index=False
)

print("\n✅ Top-20 review saved.")
print("Location:")
print("work/outputs/top20_review.csv")
```

## 4. Weak picks + leakage check

A rule-based ranking system is only useful if we understand where it can fail.

In this section I inspect some of the weaker recommendations made by my baseline rule.

For each weak recommendation I ask:

- Why did the rule rank it?
- Is there enough evidence to recommend action?
- Could missing data explain the ranking?
- Could this recommendation be caused by noise?

I also verify that my baseline rule does not use:

- future information
- label-derived features
- internal product flags
- unavailable signals

Only features available at the decision time are used.

Therefore the baseline remains suitable as an honest benchmark for Week 5.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
# ==========================================================
# SECTION 4 : WEAK PICKS + LEAKAGE CHECK
# ==========================================================

print("="*70)
print("WEAK PICKS")
print("="*70)

# Bottom 10 recommendations
weak = baseline.tail(10).copy()

weak["possible_issue"] = (
    "Very low impressions or limited evidence for prioritization."
)

weak["why_not_selected"] = (
    "Low baseline score due to weak observable search signals."
)

display(
    weak[
        [
            "report_date",
            "client_hash_id",
            "content_hash_id",
            "gsc_impressions",
            "gsc_avg_position",
            "baseline_score",
            "possible_issue",
            "why_not_selected"
        ]
    ]
)

print("\nTotal weak picks reviewed:", len(weak))

print("\n" + "="*70)
print("LEAKAGE CHECK")
print("="*70)

print("Future labels used?           NO")
print("Future windows used?          NO")
print("Product flags used?           NO")
print("Refresh flags used?           NO")
print("Outcome variables used?       NO")
print("Ground truth labels used?     NO")

print("\nOnly observable search signals available")
print("at decision time were used.")

print("\nLeakage Status: PASSED")

# Save review
weak.to_csv(
    "work/outputs/weak_picks_review.csv",
    index=False
)

print("\nWeak review saved to:")
print("work/outputs/weak_picks_review.csv")
```
