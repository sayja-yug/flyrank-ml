## 3. Success metric

## Success Metric

The primary success metric for this project is **Precision@K** because the goal is to rank webpages according to their refresh priority. Content editors can only review a limited number of pages, so the quality of the highest-ranked recommendations is more important than overall accuracy.

Precision@K measures the proportion of pages in the top-ranked list that are genuinely good candidates for review. A higher Precision@K means that editors spend more of their limited time reviewing pages that are likely to require attention, improving the efficiency of the content review process.

In addition to Precision@K, I would also consider **Average Precision** and **Recall** during later model evaluation. These metrics provide additional insight into how well the ranking identifies important pages and whether meaningful opportunities are being missed.

Overall, Precision@K best matches the business objective because the model is designed to prioritize a limited number of high-value pages for editorial review rather than make equally important predictions for every page in the dataset.

## 5. Why ML beats a fixed rule here

A simple rule, such as selecting every page with a **downward trend**, is not sufficient for this problem because it ignores many other factors that influence the priority of a content refresh. Two pages may both show a declining trend, but one may have very high search visibility and business impact while the other has very little traffic. Treating both pages equally would lead to poor prioritization.

Machine Learning is more appropriate because it can learn patterns from multiple interacting signals, including impressions, clicks, CTR, sessions, average position, content age, engagement metrics, and trend information. Instead of relying on manually written thresholds, the model can combine these features to produce a priority score that better reflects the overall refresh opportunity of each webpage.

This approach supports the business objective of ranking webpages according to their estimated value for review. It helps content editors focus on the pages that are most likely to benefit from attention while making better use of limited editorial resources. As the framing guide notes, ML is justified here because the decision depends on many interacting signals rather than a single explainable rule.
