# 1. Question

## Research question

Can historical search-performance signals from March 2026 be used to
prioritize content items for human review when identifying potential
future search-performance decline?

## Decision this research supports

The practical decision is:

> Which content items should a content or SEO team review first?

The model is therefore treated as a **decision-support ranking system**.
It produces a ranked queue so that limited human review effort can be
directed toward higher-priority items.

The decision workflow is:

**March search signals**
→ **model scoring**
→ **rank content opportunities**
→ **human review**
→ **investigate / refresh / monitor**
→ **observe later performance**

## What the research is trying to measure

The study evaluates whether March search-performance features provide useful
ranking information about an observed decline in April clicks.

The analysis uses:

- March 2026 as the decision / feature window
- April 2026 as the observed outcome window
- client × content as the unit of analysis
- March impressions, March clicks, and March average position as model inputs
- `future_decline` as the observed outcome
- the Week-4 baseline as the comparison reference

The final Week-5/Week-6 evaluation uses a client-grouped validation design
to reduce the risk that the same client appears in both training and testing.

## What this research does NOT claim

This study does **not** claim that:

- the model will predict future performance perfectly;
- the model generalizes equally to every client or future period;
- refreshing a page will cause its search performance to recover;
- the model establishes a causal relationship between content refreshes and
  search-performance improvement;
- model scores should automatically trigger content changes.

Instead, the result is interpreted as **observed and measured ranking evidence**
that can support human prioritization.

## Why this question matters

Content teams may have many pages that could potentially be reviewed, while
human review time is limited.

A useful model does not need to replace the content team. It can instead
help organize the review queue so that higher-scoring opportunities receive
attention first.

The goal of this capstone is therefore not autonomous content optimization.
The goal is to evaluate whether a validated ranking signal can provide
practical, evidence-based decision support.

## Research framing

The central question can be stated simply:

> **Can March search-performance signals help prioritize which content
> deserves human review for potential future decline?**

The answer will be evaluated using measured model performance against the
Week-4 baseline under an honest validation design.

# 3. Methodology

## 3.1 Analytical objective

The analysis evaluates whether March 2026 search-performance signals can support the prioritization of content items for human review of potential future decline.

The objective is **decision support**, not automated content optimization and not causal inference.

The analysis asks whether content can be ranked using information available during the March decision window, and whether that ranking provides useful evidence when evaluated against the subsequently observed April outcome.

---

## 3.2 Unit of analysis

The unit of analysis is:

**one client × one content item**

Each analytical row represents one unique client-content pair.

This unit is used consistently for feature construction, target definition, model evaluation, and ranking.

---

## 3.3 Decision and outcome windows

Two separate time windows are used:

- **Decision / feature window:** March 2026
- **Observed outcome window:** April 2026

March information is treated as information that would have been available when making the prioritization decision.

April information is treated as future observed information and is used only to construct/evaluate the target.

This separation is important because using April information as an input to a March decision would create temporal leakage.

---

## 3.4 Model features

The model uses three March-only features:

1. `march_impressions`
2. `march_clicks`
3. `march_avg_position`

These features summarize search-performance signals observed during March.

The model does **not** use:

- April impressions
- April clicks
- April average position
- `click_change_pct`
- `future_decline`
- `baseline_score`
- `reason_code`
- `action_label`

as model features.

The Week-4 baseline is retained as a comparison system rather than as an input feature.

---

## 3.5 Target definition

The observed target is:

`future_decline`

The target represents whether the content experienced the defined future click-decline condition in April relative to the March reference.

The target is constructed from the future outcome window and therefore must never be available to the model at decision time.

The target is used for evaluation after the March-based prediction/ranking has been produced.

---

## 3.6 Models and baseline

The Week-5 analysis evaluated:

- Logistic Regression
- Random Forest
- Week-4 baseline

The Week-4 baseline provides a pre-existing decision score against which the model-based ranking can be compared.

The machine-learning models use only the approved March features.

The final Week-7 action playbook uses a validated model score to rank items for human review.

---

## 3.7 Validation design

Two validation designs were examined:

### Random split

A conventional random train/test split was evaluated first.

The random split produced an Average Precision of:

**0.6085**

However, content observations from the same client can be related. A random split can therefore place observations from the same client in both training and testing data.

This can make the evaluation less representative of performance on previously unseen clients.

### Client-grouped split

The Week-6 audit therefore evaluated the model using a client-grouped train/test split.

Under this design:

- training clients and testing clients are separated;
- no client appears in both sets;
- the model is evaluated on clients not represented in training.

The grouped validation Average Precision was:

**0.5754**

The Week-4 baseline Average Precision on the same grouped evaluation was:

**0.5564**

Therefore, the grouped model result was:

**+0.0190 Average Precision relative to the Week-4 baseline.**

The grouped result is treated as the more conservative validation reference.

---

## 3.8 Evaluation metrics

The main ranking metric is **Average Precision (AP)** because the task is a prioritization/ranking problem involving a binary future-decline outcome.

Precision@K was also examined in Week-5 to evaluate the quality of the highest-ranked items.

The analysis does not treat a single metric as proof of business impact.

Instead, the metrics are interpreted as measured evidence about ranking quality on the evaluated test set.

---

## 3.9 Leakage controls

The methodology includes explicit leakage controls.

The following checks were performed:

- March features were separated from April outcomes.
- Future-derived variables were excluded from the model matrix.
- `future_decline` was used only as the observed target.
- Week-4 `baseline_score` was kept separate from model features.
- Client overlap was checked under grouped validation.
- The Week-6 leakage audit confirmed that the model matrix contained only the approved March features.

These checks reduce the risk that future information or evaluation outputs are accidentally used to construct the model input.

---

## 3.10 Interpretation boundary

The results should be interpreted as **observed and measured decision-support evidence**.

The analysis does not establish that:

- refreshing a page will cause recovery;
- the model causes traffic improvement;
- the model will perform equally well for every client;
- the ranked queue should be executed automatically.

The appropriate operational interpretation is:

**The validated model provides a ranking that can help humans prioritize content for investigation or refresh review, subject to human judgment and further evidence.**

## 4.1 Model performance relative to the baseline

The primary comparison is between the Week-4 baseline and the machine-learning ranking under client-grouped validation.

Average Precision (AP) is used as the main ranking metric because the objective is to prioritize content items associated with the observed future-decline outcome.

The measured Average Precision values are:

- Week-4 baseline: **0.5564**
- Logistic Regression — random split: **0.6085**
- Logistic Regression — client-grouped split: **0.5754**

The client-grouped result is treated as the more conservative validation estimate because observations from the same client are kept within the same train or test group.

Under grouped validation, the model measured **0.5754 AP**, compared with **0.5564 AP** for the Week-4 baseline. This corresponds to an observed improvement of **0.0190 AP** over the baseline on the evaluated grouped test reference.

The random-split estimate was higher at **0.6085 AP**. The difference between the random and grouped estimates was **0.0331 AP**, with grouped validation producing the lower estimate.

This difference is important because it demonstrates that validation design affects the measured performance of the model. The random split should therefore not be presented as the primary evidence of generalization.

---

## 4.2 Results interpretation

The results provide measured evidence that the model ranking performed better than the Week-4 baseline under the client-grouped evaluation used in the validation audit.

However, the magnitude of the improvement is modest:

**Grouped model AP − baseline AP = +0.0190**

The result should therefore be interpreted as evidence that the model may provide useful additional ranking signal for decision-support, rather than as proof of business impact.

The analysis does not establish that acting on the model recommendations will cause traffic recovery, increase clicks, or improve search performance.

---

## 4.3 Validation-design sensitivity

The comparison between random and grouped validation is an important result in its own right.

The random split produced an AP of **0.6085**, while the client-grouped split produced an AP of **0.5754**.

Therefore:

**Grouped AP − Random AP = −0.0331**

The lower grouped score indicates that the estimated performance depends on how the train/test data are separated.

Because the grouped split prevents the same client from appearing in both training and testing data, it provides a more conservative estimate for this analysis.

## LIMITATIONS & HONEST FRAMING

## Why this section matters

The results above provide measured evidence about the ranking model under the evaluated validation design. They do not establish that the model will produce the same performance in every future setting, nor do they establish that acting on a recommendation will cause a content item to recover.

This section makes the boundary between **measured evidence**, **decision-support interpretation**, and **claims that are not supported by the analysis** explicit.

---

## 1. Validation limitations

The strongest validation result used in this project is the client-grouped split.

The grouped evaluation is more conservative than the random split because content from the same client is kept on one side of the train/test boundary. The evaluated grouped Average Precision was **0.5754**, compared with **0.6085** for the random split.

The difference indicates that validation design affects the measured estimate.

However, the analysis uses one client-grouped train/test evaluation rather than repeated cross-validation across many independent client partitions. Therefore, the grouped result should be interpreted as an evaluated reference rather than a universal estimate of future performance.

**INVESTIGATE**

Investigation should occur before deciding whether a content action is appropriate.

A medium-priority score should therefore be interpreted as a prioritization signal rather than as a diagnosis.

Items in the monitor group are assigned:

**MONITOR**

Human reviewers can continue observing their search-performance signals and reconsider them if additional evidence appears.

```python
# ============================================================================
# CAPSTONE — SECTION 6: RANKED RECOMMENDATIONS
# ============================================================================
# Purpose:
# - Load the Week-7 action playbook.
# - Verify the ranked queue.
# - Summarize priority levels and actions.
# - Check that every recommendation requires human review.
# - Create paper-ready recommendation evidence.
# ============================================================================

import os
import pandas as pd
import numpy as np

print("=" * 75)
print("CAPSTONE — SECTION 6: RANKED RECOMMENDATIONS")
print("=" * 75)

# ---------------------------------------------------------------------------
# 1. Locate Week-7 action playbook
# ---------------------------------------------------------------------------

QUEUE_PATH = "/content/week7_ranked_action_queue.csv"
REVIEW_PATH = "/content/week7_human_review_queue.csv"
RULES_PATH = "/content/week7_human_review_rules.csv"
NOGO_PATH = "/content/week7_no_go_list.csv"

for path in [QUEUE_PATH, REVIEW_PATH, RULES_PATH, NOGO_PATH]:
    if not os.path.exists(path):
        raise FileNotFoundError(
            f"Required Week-7 artifact not found:\n{path}"
        )

queue_df = pd.read_csv(QUEUE_PATH)
review_queue = pd.read_csv(REVIEW_PATH)
review_rules = pd.read_csv(RULES_PATH)
no_go_df = pd.read_csv(NOGO_PATH)

print("\n" + "=" * 75)
print("WEEK-7 ACTION PLAYBOOK LOADED")
print("=" * 75)

print(f"Ranked queue rows: {len(queue_df)}")

# ---------------------------------------------------------------------------
# 2. Required columns
# ---------------------------------------------------------------------------

required_columns = [
    "rank",
    "client_hash_id",
    "content_hash_id",
    "logistic_score",
    "priority_band",
    "recommended_action",
    "priority_reason",
    "human_review_required",
    "automation_allowed"
]

print("\n" + "=" * 75)
print("REQUIRED COLUMN CHECK")
print("=" * 75)

missing_columns = [
    column for column in required_columns
    if column not in queue_df.columns
]

if missing_columns:
    raise ValueError(
        f"Missing required queue columns: {missing_columns}"
    )

for column in required_columns:
    print(f"✅ {column}")

# ---------------------------------------------------------------------------
# 3. Ranking integrity
# ---------------------------------------------------------------------------

queue_df = queue_df.sort_values(
    "logistic_score",
    ascending=False
).reset_index(drop=True)

queue_df["expected_rank"] = np.arange(1, len(queue_df) + 1)

rank_valid = (
    queue_df["rank"].astype(int).to_numpy()
    == queue_df["expected_rank"].to_numpy()
).all()

print("\n" + "=" * 75)
print("RANKING CHECK")
print("=" * 75)

print(f"Rows: {len(queue_df)}")
print(f"Top score: {queue_df['logistic_score'].max():.6f}")
print(f"Bottom score: {queue_df['logistic_score'].min():.6f}")

if rank_valid:
    print("✅ Rank ordering is valid.")
else:
    print("⚠️ Rank column did not match expected ordering.")

queue_df = queue_df.drop(columns=["expected_rank"])

# ---------------------------------------------------------------------------
# 4. Priority distribution
# ---------------------------------------------------------------------------

priority_counts = (
    queue_df["priority_band"]
    .value_counts()
    .reindex(
        ["HIGH_PRIORITY", "MEDIUM_PRIORITY", "MONITOR"],
        fill_value=0
    )
)

priority_distribution = pd.DataFrame({
    "priority_band": priority_counts.index,
    "rows": priority_counts.values
})

priority_distribution["percentage"] = (
    priority_distribution["rows"]
    / len(queue_df)
    * 100
)

print("\n" + "=" * 75)
print("PRIORITY DISTRIBUTION")
print("=" * 75)

print(priority_distribution.to_string(index=False))

# ---------------------------------------------------------------------------
# 5. Action distribution
# ---------------------------------------------------------------------------

action_distribution = (
    queue_df["recommended_action"]
    .value_counts()
    .reset_index()
)

action_distribution.columns = [
    "recommended_action",
    "rows"
]

action_distribution["percentage"] = (
    action_distribution["rows"]
    / len(queue_df)
    * 100
)

print("\n" + "=" * 75)
print("ACTION DISTRIBUTION")
print("=" * 75)

print(action_distribution.to_string(index=False))

# ---------------------------------------------------------------------------
# 6. Human-review safety
# ---------------------------------------------------------------------------

human_review_count = (
    queue_df["human_review_required"]
    .astype(bool)
    .sum()
)

automatic_allowed_count = (
    queue_df["automation_allowed"]
    .astype(bool)
    .sum()
)

print("\n" + "=" * 75)
print("HUMAN REVIEW SAFETY")
print("=" * 75)

print(
    f"Rows requiring human review: "
    f"{human_review_count} / {len(queue_df)}"
)

print(
    f"Rows allowed for automatic execution: "
    f"{automatic_allowed_count}"
)

if human_review_count == len(queue_df):
    print("✅ Every recommendation requires human review.")
else:
    print("⚠️ Some recommendations do not require human review.")

if automatic_allowed_count == 0:
    print("✅ Automatic execution is disabled.")
else:
    print("⚠️ Some recommendations allow automatic execution.")

# ---------------------------------------------------------------------------
# 7. No-go policy
# ---------------------------------------------------------------------------

print("\n" + "=" * 75)
print("NO-GO POLICY")
print("=" * 75)

print(f"No-go rules: {len(no_go_df)}")

if "allowed" in no_go_df.columns:
    allowed_values = (
        no_go_df["allowed"]
        .astype(str)
        .str.lower()
        .isin(["true", "1", "yes"])
    )

    if allowed_values.sum() == 0:
        print("✅ All no-go actions remain disabled.")
    else:
        print("⚠️ One or more no-go actions appear enabled.")

# ---------------------------------------------------------------------------
# 8. Build paper-ready recommendation summary
# ---------------------------------------------------------------------------

recommendation_summary = pd.DataFrame([
    {
        "priority_band": "HIGH_PRIORITY",
        "recommended_action": "REFRESH_REVIEW",
        "queue_rows": int(
            priority_counts.get("HIGH_PRIORITY", 0)
        ),
        "percentage": round(
            priority_counts.get("HIGH_PRIORITY", 0)
            / len(queue_df) * 100,
            2
        ),
        "human_review_required": True,
        "automation_allowed": False,
        "interpretation": (
            "Review first and determine whether a refresh "
            "is justified by additional evidence."
        )
    },
    {
        "priority_band": "MEDIUM_PRIORITY",
        "recommended_action": "INVESTIGATE",
        "queue_rows": int(
            priority_counts.get("MEDIUM_PRIORITY", 0)
        ),
        "percentage": round(
            priority_counts.get("MEDIUM_PRIORITY", 0)
            / len(queue_df) * 100,
            2
        ),
        "human_review_required": True,
        "automation_allowed": False,
        "interpretation": (
            "Investigate the evidence before deciding "
            "whether further action is warranted."
        )
    },
    {
        "priority_band": "MONITOR",
        "recommended_action": "MONITOR",
        "queue_rows": int(
            priority_counts.get("MONITOR", 0)
        ),
        "percentage": round(
            priority_counts.get("MONITOR", 0)
            / len(queue_df) * 100,
            2
        ),
        "human_review_required": True,
        "automation_allowed": False,
        "interpretation": (
            "Continue observation unless additional "
            "evidence changes the priority."
        )
    }
])

# ---------------------------------------------------------------------------
# 9. Top-ranked paper review sample
# ---------------------------------------------------------------------------

top_review_sample = queue_df[
    [
        "rank",
        "client_hash_id",
        "content_hash_id",
        "logistic_score",
        "priority_band",
        "recommended_action",
        "priority_reason",
        "human_review_required",
        "automation_allowed"
    ]
].head(20).copy()

print("\n" + "=" * 75)
print("TOP 20 PAPER REVIEW SAMPLE")
print("=" * 75)

print(top_review_sample.to_string(index=False))

# ---------------------------------------------------------------------------
# 10. Export recommendation artifacts
# ---------------------------------------------------------------------------

OUTPUT_DIR = "/content/work/outputs"
os.makedirs(OUTPUT_DIR, exist_ok=True)

summary_path = os.path.join(
    OUTPUT_DIR,
    "capstone_ranked_recommendations.csv"
)

priority_path = os.path.join(
    OUTPUT_DIR,
    "capstone_recommendation_priority_distribution.csv"
)

sample_path = os.path.join(
    OUTPUT_DIR,
    "capstone_top20_recommendation_sample.csv"
)

recommendation_summary.to_csv(
    summary_path,
    index=False
)

priority_distribution.to_csv(
    priority_path,
    index=False
)

top_review_sample.to_csv(
    sample_path,
    index=False
)

# ---------------------------------------------------------------------------
# 11. Final Section 6 checks
# ---------------------------------------------------------------------------

print("\n" + "=" * 75)
print("SECTION 6 FINAL CHECK")
print("=" * 75)

print(f"Total ranked items: {len(queue_df)}")
print(
    f"High priority: "
    f"{priority_counts.get('HIGH_PRIORITY', 0)}"
)
print(
    f"Medium priority: "
    f"{priority_counts.get('MEDIUM_PRIORITY', 0)}"
)
print(
    f"Monitor: "
    f"{priority_counts.get('MONITOR', 0)}"
)

if (
    priority_counts.sum() == len(queue_df)
    and human_review_count == len(queue_df)
    and automatic_allowed_count == 0
):
    print("✅ All queue items have a valid priority.")
    print("✅ Every recommendation requires human review.")
    print("✅ Automatic execution is disabled.")
else:
    raise ValueError(
        "Section 6 safety or priority checks failed."
    )

print("\n" + "=" * 75)
print("SECTION 6 ARTIFACTS")
print("=" * 75)

print(f"✅ {summary_path}")
print(f"✅ {priority_path}")
print(f"✅ {sample_path}")

print("\n" + "=" * 75)
print("SECTION 6 COMPLETE")
print("=" * 75)

print("""
Paper interpretation:

The ranked queue is a decision-support artifact. Higher-ranked
content receives earlier human attention, but model rank alone
does not establish that a content intervention will improve
future performance.

Recommended workflow:

MODEL SCORE
     ↓
RANK
     ↓
PRIORITY BAND
     ↓
HUMAN REVIEW
     ↓
INVESTIGATE / REFRESH REVIEW / MONITOR
""")
```

# 7. Artifacts the Paper Embeds

This section identifies the evidence produced by the capstone analysis that will be reused in the final research paper.

The purpose is not to expose the full analytical dataset or individual content records. Instead, the paper will embed public-safe aggregate evidence that supports the methodology, results, limitations, and ranked-recommendation sections.

## Paper-facing evidence

The final paper will use the following evidence:

1. **Validation comparison**
   - Week-4 baseline Average Precision
   - Week-5 Logistic Regression under a random split
   - Week-5 Logistic Regression under client-grouped validation
   - The grouped result is treated as the more conservative validation reference.

2. **Validation-design figure**
   - Shows the difference between random-split and client-grouped evaluation.
   - Supports the methodological discussion about why grouped validation is important.

3. **Model-vs-baseline result**
   - Compares the Week-5 grouped model with the Week-4 baseline on the evaluated reference.
   - The result is presented as measured ranking evidence, not causal business impact.

4. **Ranked recommendation summary**
   - High-priority: 22 items
   - Medium-priority: 31 items
   - Monitor: 53 items
   - These counts describe the decision-support queue without exposing individual client or content identifiers.

5. **Human-review policy**
   - Every recommendation requires human review.
   - Automatic execution is disabled.
   - The model is used to prioritize attention rather than execute content changes.

6. **Monitoring and retraining policy**
   - Monitor data quality, score distribution, priority distribution, observed performance, human-review overrides, and changes in the target/business definition.
   - Monitoring triggers investigation.
   - Retraining is not automatic.

## Public-safety boundary

The final paper must not publish:

- client identifiers;
- content identifiers;
- private URLs;
- private search queries;
- individual customer records;
- authentication tokens or secrets;
- raw production-level records that are not intended for public release.

The paper should use aggregate statistics, methodology descriptions, public-safe figures, and reproducibility receipts instead.

## Evidence hierarchy

The paper should distinguish between:

**Observed / measured evidence**
- model metrics;
- validation results;
- queue counts;
- feature definitions;
- error counts.

**Decision-support interpretation**
- which content should receive earlier human attention;
- how the queue can support review;
- how monitoring can inform future investigation.

**Claims that are not supported**
- the model causes traffic recovery;
- refreshing a page guarantees improvement;
- the model generalizes to every client or future dataset;
- the model can safely make content changes autonomously.

## Reproducibility

The capstone evidence is connected to the previous project stages:

- Week 4 → baseline;
- Week 5 → model development and model-vs-baseline comparison;
- Week 6 → grouped validation, leakage audit, error analysis, and claim rewriting;
- Week 7 → ranked action playbook, human-review policy, no-go rules, and monitoring triggers;
- Week 8 → research-paper assembly and deployment.

The final paper therefore uses the capstone notebook as the synthesis layer rather than treating the paper as a separate analysis.

## Final paper evidence principle

Every important claim in the paper should be traceable to either:

1. a saved metric or figure,
2. a documented methodological decision, or
3. an explicitly stated limitation.

The paper should present the work as an evaluated decision-support workflow, not as an autonomous production system.

```python
# ============================================================================
# CAPSTONE — SECTION 7: ARTIFACTS THE PAPER EMBEDS
# ============================================================================
# This section verifies that the research-paper artifacts exist and that
# the detailed Week-7 ranked queue is the correct 106-row artifact.
#
# IMPORTANT:
# The full Week-7 queue is stored directly under /content/.
#
# We DO NOT use:
# /content/work/outputs/capstone_ranked_recommendations.csv
#
# because that file is only the 3-row aggregate priority summary.
# ============================================================================

import os
import pandas as pd

print("=" * 75)
print("CAPSTONE — SECTION 7: ARTIFACTS THE PAPER EMBEDS")
print("=" * 75)


# ============================================================================
# 1. DEFINE ARTIFACT PATHS
# ============================================================================

OUTPUT_DIR = "/content/work/outputs"
FIGURE_DIR = "/content/work/figures"

# ---------------------------------------------------------------------------
# Detailed Week-7 ranked queue
# ---------------------------------------------------------------------------

WEEK7_QUEUE_PATH = "/content/week7_ranked_action_queue.csv"

# ---------------------------------------------------------------------------
# Week-7 supporting artifacts
# ---------------------------------------------------------------------------

WEEK7_REVIEW_PATH = "/content/week7_human_review_rules.csv"
WEEK7_NOGO_PATH = "/content/week7_no_go_list.csv"
WEEK7_MONITORING_PATH = "/content/week7_monitoring_triggers.csv"
WEEK7_MONITORING_REFERENCE_PATH = "/content/week7_monitoring_reference.csv"

# ---------------------------------------------------------------------------
# Capstone artifacts
# ---------------------------------------------------------------------------

CAPSTONE_DATA_AUDIT = os.path.join(
    OUTPUT_DIR,
    "capstone_data_audit.csv"
)

CAPSTONE_METHODOLOGY = os.path.join(
    OUTPUT_DIR,
    "capstone_methodology_receipt.csv"
)

CAPSTONE_RESULTS = os.path.join(
    OUTPUT_DIR,
    "capstone_results_comparison.csv"
)

CAPSTONE_RESULTS_RECEIPT = os.path.join(
    OUTPUT_DIR,
    "capstone_results_receipt.csv"
)

CAPSTONE_LIMITATIONS = os.path.join(
    OUTPUT_DIR,
    "capstone_limitations_receipt.csv"
)

CAPSTONE_CLAIMS = os.path.join(
    OUTPUT_DIR,
    "capstone_claim_boundary_audit.csv"
)

CAPSTONE_TOP20 = os.path.join(
    OUTPUT_DIR,
    "capstone_top20_recommendation_sample.csv"
)

CAPSTONE_PRIORITY = os.path.join(
    OUTPUT_DIR,
    "capstone_recommendation_priority_distribution.csv"
)

CAPSTONE_QUEUE_SUMMARY = os.path.join(
    OUTPUT_DIR,
    "capstone_ranked_recommendations.csv"
)

RESULTS_FIGURE_1 = os.path.join(
    FIGURE_DIR,
    "capstone_results_average_precision.png"
)

RESULTS_FIGURE_2 = os.path.join(
    FIGURE_DIR,
    "capstone_validation_design_effect.png"
)


# ============================================================================
# 2. DIRECTORY CHECK
# ============================================================================

print("\n" + "=" * 75)
print("ARTIFACT DIRECTORY CHECK")
print("=" * 75)

print("Outputs :", OUTPUT_DIR)
print("Figures :", FIGURE_DIR)

if not os.path.exists(OUTPUT_DIR):
    raise FileNotFoundError(
        f"Capstone output directory not found:\n{OUTPUT_DIR}"
    )

if not os.path.exists(FIGURE_DIR):
    print(
        "⚠️ Figures directory does not exist yet."
    )

print("✅ Output directory available.")


# ============================================================================
# 3. LOAD THE FULL WEEK-7 RANKED QUEUE
# ============================================================================

print("\n" + "=" * 75)
print("WEEK-7 RANKED QUEUE")
print("=" * 75)

if not os.path.exists(WEEK7_QUEUE_PATH):
    raise FileNotFoundError(
        "Full Week-7 ranked queue was not found:\n"
        + WEEK7_QUEUE_PATH
    )

paper_queue = pd.read_csv(
    WEEK7_QUEUE_PATH
)

print("Source:")
print(WEEK7_QUEUE_PATH)

print("\nRows:", len(paper_queue))

print("\nColumns:")
print(list(paper_queue.columns))


# ============================================================================
# 4. REQUIRED QUEUE COLUMN CHECK
# ============================================================================

required_queue_columns = [
    "rank",
    "client_hash_id",
    "content_hash_id",
    "logistic_score",
    "priority_band",
    "recommended_action",
    "priority_reason",
    "human_review_required",
    "automation_allowed"
]

print("\n" + "=" * 75)
print("RANKED QUEUE COLUMN CHECK")
print("=" * 75)

missing_queue_columns = [
    column
    for column in required_queue_columns
    if column not in paper_queue.columns
]

if missing_queue_columns:
    raise ValueError(
        "The Week-7 ranked queue is missing required columns:\n"
        + str(missing_queue_columns)
    )

for column in required_queue_columns:
    print("✅", column)


# ============================================================================
# 5. QUEUE SIZE CHECK
# ============================================================================

print("\n" + "=" * 75)
print("QUEUE SIZE CHECK")
print("=" * 75)

print("Queue rows:", len(paper_queue))

if len(paper_queue) != 106:
    raise ValueError(
        "Expected 106 Week-7 ranked queue rows, "
        f"but found {len(paper_queue)}."
    )

print("✅ Full 106-row Week-7 queue confirmed.")


# ============================================================================
# 6. RANKING CHECK
# ============================================================================

print("\n" + "=" * 75)
print("RANKING CHECK")
print("=" * 75)

paper_queue = paper_queue.sort_values(
    "rank"
).reset_index(drop=True)

expected_ranks = list(
    range(1, len(paper_queue) + 1)
)

actual_ranks = paper_queue["rank"].tolist()

if actual_ranks != expected_ranks:
    raise ValueError(
        "Rank sequence is not continuous from 1 to queue size."
    )

print("First rank:", paper_queue["rank"].min())
print("Last rank :", paper_queue["rank"].max())

print("Top score :", paper_queue["logistic_score"].max())
print("Bottom score:", paper_queue["logistic_score"].min())

print("✅ Rank sequence is valid.")


# ============================================================================
# 7. IDENTIFIER CHECK
# ============================================================================

print("\n" + "=" * 75)
print("IDENTIFIER CHECK")
print("=" * 75)

missing_client_ids = paper_queue[
    "client_hash_id"
].isna().sum()

missing_content_ids = paper_queue[
    "content_hash_id"
].isna().sum()

duplicate_pairs = paper_queue.duplicated(
    subset=[
        "client_hash_id",
        "content_hash_id"
    ]
).sum()

print("Missing client IDs :", missing_client_ids)
print("Missing content IDs:", missing_content_ids)
print("Duplicate client-content pairs:", duplicate_pairs)

if missing_client_ids > 0:
    raise ValueError(
        "Some client_hash_id values are missing."
    )

if missing_content_ids > 0:
    raise ValueError(
        "Some content_hash_id values are missing."
    )

if duplicate_pairs > 0:
    raise ValueError(
        "Duplicate client-content pairs found in ranked queue."
    )

print("✅ Identifiers are present.")
print("✅ No duplicate client-content pairs.")


# ============================================================================
# 8. MODEL SCORE CHECK
# ============================================================================

print("\n" + "=" * 75)
print("MODEL SCORE CHECK")
print("=" * 75)

missing_scores = paper_queue[
    "logistic_score"
].isna().sum()

infinite_scores = (
    ~paper_queue["logistic_score"].map(
        lambda x: pd.notna(x) and x != float("inf") and x != float("-inf")
    )
).sum()

print("Missing scores  :", missing_scores)
print("Invalid scores  :", infinite_scores)

if missing_scores > 0:
    raise ValueError(
        "Missing logistic_score values found."
    )

if infinite_scores > 0:
    raise ValueError(
        "Invalid/infinite logistic_score values found."
    )

print("Minimum score:", paper_queue["logistic_score"].min())
print("Maximum score:", paper_queue["logistic_score"].max())

print("✅ Model scores are valid.")


# ============================================================================
# 9. HUMAN REVIEW / AUTOMATION CHECK
# ============================================================================

print("\n" + "=" * 75)
print("HUMAN REVIEW SAFETY CHECK")
print("=" * 75)

human_review_count = (
    paper_queue["human_review_required"]
    .astype(bool)
    .sum()
)

automation_allowed_count = (
    paper_queue["automation_allowed"]
    .astype(bool)
    .sum()
)

print(
    "Rows requiring human review:",
    human_review_count,
    "/",
    len(paper_queue)
)

print(
    "Rows allowed for automatic execution:",
    automation_allowed_count
)

if human_review_count != len(paper_queue):
    raise ValueError(
        "Not every queue item requires human review."
    )

if automation_allowed_count != 0:
    raise ValueError(
        "Some queue items are marked for automatic execution."
    )

print("✅ Every recommendation requires human review.")
print("✅ Automatic execution is disabled.")


# ============================================================================
# 10. PRIORITY DISTRIBUTION
# ============================================================================

print("\n" + "=" * 75)
print("PRIORITY DISTRIBUTION")
print("=" * 75)

priority_distribution = (
    paper_queue["priority_band"]
    .value_counts()
    .rename_axis("priority_band")
    .reset_index(name="rows")
)

priority_distribution["percentage"] = (
    priority_distribution["rows"]
    / len(paper_queue)
    * 100
)

print(
    priority_distribution.to_string(
        index=False
    )
)

priority_total = priority_distribution["rows"].sum()

if priority_total != len(paper_queue):
    raise ValueError(
        "Priority distribution does not account for all queue rows."
    )

print("✅ Priority distribution accounts for every queue item.")


# ============================================================================
# 11. ACTION DISTRIBUTION
# ============================================================================

print("\n" + "=" * 75)
print("ACTION DISTRIBUTION")
print("=" * 75)

action_distribution = (
    paper_queue["recommended_action"]
    .value_counts()
    .rename_axis("recommended_action")
    .reset_index(name="rows")
)

action_distribution["percentage"] = (
    action_distribution["rows"]
    / len(paper_queue)
    * 100
)

print(
    action_distribution.to_string(
        index=False
    )
)

print("✅ Recommended actions are present.")


# ============================================================================
# 12. SUPPORTING WEEK-7 ARTIFACT CHECK
# ============================================================================

print("\n" + "=" * 75)
print("WEEK-7 SUPPORTING ARTIFACT CHECK")
print("=" * 75)

week7_supporting_files = {
    "Human-review rules": WEEK7_REVIEW_PATH,
    "No-go list": WEEK7_NOGO_PATH,
    "Monitoring triggers": WEEK7_MONITORING_PATH,
    "Monitoring reference": WEEK7_MONITORING_REFERENCE_PATH
}

for label, path in week7_supporting_files.items():

    if os.path.exists(path):

        df = pd.read_csv(path)

        print(
            f"✅ {label}: {len(df)} rows"
        )

    else:

        print(
            f"⚠️ {label} not found: {path}"
        )


# ============================================================================
# 13. CAPSTONE ARTIFACT CHECK
# ============================================================================

print("\n" + "=" * 75)
print("CAPSTONE ARTIFACT CHECK")
print("=" * 75)

capstone_artifacts = {
    "Data audit": CAPSTONE_DATA_AUDIT,
    "Methodology receipt": CAPSTONE_METHODOLOGY,
    "Results comparison": CAPSTONE_RESULTS,
    "Results receipt": CAPSTONE_RESULTS_RECEIPT,
    "Limitations receipt": CAPSTONE_LIMITATIONS,
    "Claim boundary audit": CAPSTONE_CLAIMS,
    "Top-20 recommendation sample": CAPSTONE_TOP20,
    "Priority distribution": CAPSTONE_PRIORITY,
    "Recommendation summary": CAPSTONE_QUEUE_SUMMARY
}

artifact_status = []

for label, path in capstone_artifacts.items():

    exists = os.path.exists(path)

    artifact_status.append(
        {
            "artifact": label,
            "path": path,
            "exists": exists
        }
    )

    if exists:
        print("✅", label)
    else:
        print("⚠️", label, "NOT FOUND")


# ============================================================================
# 14. FIGURE CHECK
# ============================================================================

print("\n" + "=" * 75)
print("PAPER FIGURE CHECK")
print("=" * 75)

figure_paths = {
    "Average Precision comparison": RESULTS_FIGURE_1,
    "Validation design effect": RESULTS_FIGURE_2
}

for label, path in figure_paths.items():

    if os.path.exists(path):
        print("✅", label)
    else:
        print("⚠️", label, "NOT FOUND")


# ============================================================================
# 15. RESULTS EVIDENCE CHECK
# ============================================================================

print("\n" + "=" * 75)
print("VALIDATION EVIDENCE CHECK")
print("=" * 75)

if os.path.exists(CAPSTONE_RESULTS):

    results_df = pd.read_csv(
        CAPSTONE_RESULTS
    )

    print(
        results_df.to_string(
            index=False
        )
    )

    print("✅ Results comparison is available.")

else:

    print(
        "⚠️ Results comparison file not found."
    )


# ============================================================================
# 16. TOP-20 PAPER SAMPLE CHECK
# ============================================================================

print("\n" + "=" * 75)
print("TOP-20 PAPER SAMPLE CHECK")
print("=" * 75)

if os.path.exists(CAPSTONE_TOP20):

    top20_df = pd.read_csv(
        CAPSTONE_TOP20
    )

    print(
        "Top-20 sample rows:",
        len(top20_df)
    )

    required_top20_columns = [
        "rank",
        "logistic_score",
        "priority_band",
        "recommended_action",
        "priority_reason",
        "human_review_required",
        "automation_allowed"
    ]

    missing_top20_columns = [
        c
        for c in required_top20_columns
        if c not in top20_df.columns
    ]

    if missing_top20_columns:

        print(
            "⚠️ Missing top-20 columns:",
            missing_top20_columns
        )

    else:

        print("✅ Top-20 recommendation sample is available.")

else:

    print(
        "⚠️ Top-20 recommendation sample not found."
    )


# ============================================================================
# 17. FINAL ARTIFACT INVENTORY
# ============================================================================

print("\n" + "=" * 75)
print("FINAL PAPER ARTIFACT INVENTORY")
print("=" * 75)

print("\nPrimary detailed queue:")
print("✅", WEEK7_QUEUE_PATH)

print("\nSupporting Week-7 artifacts:")

for label, path in week7_supporting_files.items():

    if os.path.exists(path):
        print("✅", path)

print("\nCapstone artifacts:")

for label, path in capstone_artifacts.items():

    if os.path.exists(path):
        print("✅", path)

print("\nFigures:")

for label, path in figure_paths.items():

    if os.path.exists(path):
        print("✅", path)


# ============================================================================
# 18. PAPER INTERPRETATION
# ============================================================================

print("\n" + "=" * 75)
print("PAPER INTERPRETATION")
print("=" * 75)

print(
    """
The Week-7 ranked queue is treated as a decision-support artifact.

Higher-ranked items receive earlier human attention based on the
evaluated model score and priority rules.

The ranking does not prove that a content intervention will improve
future performance and should not be interpreted as a causal result.

The paper therefore presents the queue together with:

    model evaluation
        ↓
    validation design
        ↓
    limitations
        ↓
    human-review policy
        ↓
    monitoring framework

The model is decision-support, not an autonomous content system.
"""
)


# ============================================================================
# 19. SAVE A SECTION-7 ARTIFACT RECEIPT
# ============================================================================

section7_receipt = pd.DataFrame(
    [
        {
            "artifact": "week7_ranked_action_queue",
            "path": WEEK7_QUEUE_PATH,
            "rows": len(paper_queue),
            "status": "verified"
        },
        {
            "artifact": "week7_human_review_rules",
            "path": WEEK7_REVIEW_PATH,
            "rows": (
                len(pd.read_csv(WEEK7_REVIEW_PATH))
                if os.path.exists(WEEK7_REVIEW_PATH)
                else 0
            ),
            "status": (
                "verified"
                if os.path.exists(WEEK7_REVIEW_PATH)
                else "missing"
            )
        },
        {
            "artifact": "week7_no_go_list",
            "path": WEEK7_NOGO_PATH,
            "rows": (
                len(pd.read_csv(WEEK7_NOGO_PATH))
                if os.path.exists(WEEK7_NOGO_PATH)
                else 0
            ),
            "status": (
                "verified"
                if os.path.exists(WEEK7_NOGO_PATH)
                else "missing"
            )
        },
        {
            "artifact": "week7_monitoring_triggers",
            "path": WEEK7_MONITORING_PATH,
            "rows": (
                len(pd.read_csv(WEEK7_MONITORING_PATH))
                if os.path.exists(WEEK7_MONITORING_PATH)
                else 0
            ),
            "status": (
                "verified"
                if os.path.exists(WEEK7_MONITORING_PATH)
                else "missing"
            )
        }
    ]
)

SECTION7_RECEIPT_PATH = os.path.join(
    OUTPUT_DIR,
    "capstone_artifacts_receipt.csv"
)

section7_receipt.to_csv(
    SECTION7_RECEIPT_PATH,
    index=False
)

print("\n" + "=" * 75)
print("SECTION 7 COMPLETE")
print("=" * 75)

print("✅ Full Week-7 ranked queue verified.")
print("✅ Queue contains 106 ranked items.")
print("✅ Human review controls verified.")
print("✅ Supporting Week-7 artifacts checked.")
print("✅ Capstone evidence artifacts checked.")
print("✅ Paper figures checked.")
print("✅ Section-7 artifact receipt saved:")

print(
    SECTION7_RECEIPT_PATH
)

print("\nPaper artifact chain is ready for final assembly.")
```
