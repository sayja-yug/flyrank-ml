## 1. Unit of analysis + time window
Unit of Analysis

For the Refresh / Content Opportunity Scoring lane, one row represents the daily performance of a single content page. Each row contains measurements such as search traffic, clicks, impressions, rankings, and other SEO-related signals for one content item on one specific date. This daily grain allows the model to evaluate how the performance of each page changes over time and identify which pages are the strongest candidates for a content refresh.

Time Window

For feature development and analysis, I will use a mid-panel month (March 2026) instead of the final month in the dataset. Using a mid-panel month reflects the information that would have been available at the decision time and avoids introducing future information into the training process. The final month (June 2026) will be treated as a held-out period and not used for developing labels or features.

```python
# ============================================================
# Install required packages
# ============================================================
!pip -q install datasets huggingface_hub pyarrow pandas

# ============================================================
# Import libraries
# ============================================================
from google.colab import userdata
from datasets import load_dataset
import pandas as pd

# ============================================================
# Read Hugging Face Token from Colab Secrets
# ============================================================
HF_TOKEN = userdata.get("HF_TOKEN")

if HF_TOKEN is None:
    raise ValueError("HF_TOKEN Secret not found. Please add it in Colab Secrets.")

print("✅ Hugging Face Token Loaded Successfully")

# ============================================================
# Load FlyRank Warehouse Dataset
# ============================================================
dataset = load_dataset(
    "FlyRank/internship-warehouse",
    "fact_content_daily_performance",
    split="train",
    token=HF_TOKEN
)

print("✅ Dataset Downloaded")

# ============================================================
# Convert to Pandas DataFrame
# ============================================================
df = dataset.to_pandas()

print("Original Shape:", df.shape)

# ============================================================
# Convert report_date to datetime
# ============================================================
df["report_date"] = pd.to_datetime(df["report_date"])

# ============================================================
# Filter to March 2026 (Mid-panel Month)
# ============================================================
march_df = df[
    (df["report_date"] >= "2026-03-01") &
    (df["report_date"] <= "2026-03-31")
].copy()

print("\n✅ March 2026 Data Loaded")
print("Shape:", march_df.shape)

# ============================================================
# Use March data as the working dataframe
# ============================================================
df = march_df

# ============================================================
# Preview Dataset
# ============================================================
display(df.head())

print("\nColumns:")
print(df.columns.tolist())

print("\nDate Range:")
print(df["report_date"].min(), "to", df["report_date"].max())

print("\nUnique Clients :", df["client_hash_id"].nunique())
print("Unique Content Pages :", df["content_hash_id"].nunique())
```

## 2. Fields: feature / label / context / excluded

## Feature

Features are the variables that are available at the moment a decision is made about improving a content page. These variables describe the current SEO performance and user engagement of the page.

Features used:

- gsc_impressions
- gsc_clicks
- gsc_avg_position
- gsc_sum_position
- ga4_pageviews
- ga4_sessions
- ga4_users
- ga4_engaged_sessions
- ga4_total_engagement_sec
- sessions_organic
- sessions_direct
- sessions_referral
- sessions_social
- sessions_paid
- sessions_ai
- scroll_events

These values are already known before making an optimization decision and therefore can safely be used as input features.

---

## Label

The label is the value the model will predict in the future.

For this project, I will predict future organic performance of a content page using historical Search Console and GA4 data.

Possible prediction target:

- Future Organic Sessions (sessions_organic)
or
- Future Google Search Clicks (gsc_clicks)

The label will always represent a future outcome rather than current performance.

---

## Context

Context fields identify the observation but should not directly influence the prediction.

Context fields:

- report_date
- client_hash_id
- content_hash_id
- client_has_gsc
- client_has_ga4
- gsc_data_available
- ga4_data_available

These columns describe when and where the data comes from.

---

## Excluded

Excluded fields are values that would create leakage or are unavailable when the prediction is made.

Examples include:

- Future month's clicks
- Future month's sessions
- Future engagement values
- Any information collected after the prediction date

These fields are intentionally excluded to prevent data leakage and ensure the model only learns from information available at decision time.

```python
# ==============================
# SECTION 2 - Verify Field Groups
# ==============================

features = [
    "gsc_impressions",
    "gsc_clicks",
    "gsc_sum_position",
    "gsc_avg_position",
    "ga4_pageviews",
    "ga4_sessions",
    "ga4_users",
    "ga4_engaged_sessions",
    "ga4_total_engagement_sec",
    "sessions_organic",
    "sessions_direct",
    "sessions_referral",
    "sessions_social",
    "sessions_paid",
    "sessions_ai",
    "scroll_events"
]

context = [
    "report_date",
    "client_hash_id",
    "content_hash_id",
    "client_has_gsc",
    "client_has_ga4",
    "gsc_data_available",
    "ga4_data_available"
]

print("=" * 60)
print("FEATURE COLUMNS")
print("=" * 60)
print(features)

print("\nTotal Features:", len(features))

print("\n" + "=" * 60)
print("CONTEXT COLUMNS")
print("=" * 60)
print(context)

print("\nTotal Context Columns:", len(context))

print("\n" + "=" * 60)
print("Checking Columns")
print("=" * 60)

for col in features + context:
    print(f"{col:35} {'✅ Exists' if col in df.columns else '❌ Missing'}")
```

## 3. Verify it with queries (grain, counts, missing values, windows)
### Query 1 – Verify the Grain of the Dataset

The unit of analysis defined in Section 1 is:

**One row represents one content page for one client on one reporting day.**

To verify this assumption, I check whether the combination of:

- report_date
- client_hash_id
- content_hash_id

is unique for every row.

If duplicate combinations exist, then my assumed grain is incorrect.

If no duplicates exist, then the dataset follows the expected grain.

### Query 2 – Verify Row Count and Date Span

The assignment specifies working on a single mid-panel month (March 2026).

This query verifies that the working dataset contains only the intended time window.

It reports:

- Total number of rows
- Earliest reporting date
- Latest reporting date

This confirms that all later feature engineering is performed only on the selected month.

### Query 3 – Verify Data Availability

Not every row in the warehouse contains usable Search Console (GSC) or Google Analytics 4 (GA4) data.

Before building features, I verify how many rows remain after filtering for rows where both GSC and GA4 data are available.

This ensures that later feature engineering is performed only on valid observations rather than missing or incomplete records.

```python
# =====================================================
# QUERY 1 : Verify Dataset Grain
# =====================================================

grain_columns = [
    "report_date",
    "client_hash_id",
    "content_hash_id"
]

# Total rows
total_rows = len(df)

# Unique grain combinations
unique_rows = df[grain_columns].drop_duplicates().shape[0]

# Duplicate rows
duplicate_rows = total_rows - unique_rows

print("=" * 60)
print("VERIFYING DATASET GRAIN")
print("=" * 60)

print(f"Total Rows                : {total_rows:,}")
print(f"Unique Grain Combinations : {unique_rows:,}")
print(f"Duplicate Rows            : {duplicate_rows:,}")

print()

if duplicate_rows == 0:
    print("✅ VERIFIED")
    print("Each row is uniquely identified by:")
    print("report_date + client_hash_id + content_hash_id")
else:
    print("❌ WARNING")
    print("Duplicate grain found.")
    print("The unit of analysis is NOT unique.")

# Show duplicate rows if they exist
if duplicate_rows > 0:
    print("\nDuplicate Examples:\n")
    display(
        df[df.duplicated(subset=grain_columns, keep=False)]
        .sort_values(grain_columns)
        .head(10)
    )


# =====================================================
# QUERY 2 : Verify Row Count and Date Span
# =====================================================

import pandas as pd

# Make sure report_date is datetime
df["report_date"] = pd.to_datetime(df["report_date"])

print("=" * 60)
print("VERIFYING WORKING TIME WINDOW")
print("=" * 60)

print(f"Total Rows        : {len(df):,}")

print()

print(f"Earliest Date     : {df['report_date'].min().date()}")

print(f"Latest Date       : {df['report_date'].max().date()}")

print()

print("Number of Unique Days :", df["report_date"].nunique())

print()

print("=" * 60)
print("Rows Per Date")
print("=" * 60)

display(
    df.groupby("report_date")
      .size()
      .reset_index(name="row_count")
)

# =====================================================
# QUERY 3 : Verify Data Availability
# =====================================================

print("=" * 60)
print("VERIFYING DATA AVAILABILITY")
print("=" * 60)

print("\nTotal Rows Before Filtering :", len(df))

# Keep only rows where BOTH data sources exist
available_df = df[
    (df["gsc_data_available"] == True) &
    (df["ga4_data_available"] == True)
]

print("\nTotal Rows After Filtering :", len(available_df))

print("\nRows Removed :", len(df) - len(available_df))

print("\nPercentage Remaining : {:.2f}%".format(
    (len(available_df) / len(df)) * 100
))

print("\nSample of Available Data")

display(
    available_df[
        [
            "report_date",
            "client_hash_id",
            "content_hash_id",
            "gsc_data_available",
            "ga4_data_available"
        ]
    ].head()
)
```

## 4. Data limits
## Five Features for the Refresh Opportunity Model

### Feature 1 — gsc_impressions

Reason:
This feature measures how frequently a content page appears in Google Search results. It is available before making any refresh decision and helps identify pages with strong search visibility.

Available at decision time:
Yes. Search impressions are historical observations collected before the decision.

---

### Feature 2 — gsc_clicks

Reason:
This feature measures how many users clicked the page from Google Search. It reflects current user interest and content performance.

Available at decision time:
Yes. Click counts are already known when deciding whether to refresh the page.

---

### Feature 3 — gsc_avg_position

Reason:
Average search position indicates how well the page ranks in Google Search. Pages with poor rankings but high impressions may represent optimization opportunities.

Available at decision time:
Yes. Ranking information is available before the editor decides to update the content.

---

### Feature 4 — ga4_sessions

Reason:
Sessions measure how much traffic actually reaches the page. This complements Search Console metrics by showing user visits after clicking.

Available at decision time:
Yes. Historical GA4 sessions are available before the refresh decision.

---

### Feature 5 — scroll_events

Reason:
Scroll events provide a simple measure of user engagement with the content. Pages with poor engagement may benefit from content improvements.

Available at decision time:
Yes. Scroll activity is recorded before any future optimization work.

```python
# =====================================================
# SECTION 4 : Feature Frame
# =====================================================

feature_columns = [
    "gsc_impressions",
    "gsc_clicks",
    "gsc_avg_position",
    "ga4_sessions",
    "scroll_events"
]

feature_df = df[feature_columns].copy()

print("=" * 60)
print("FEATURE FRAME")
print("=" * 60)

print("Shape:", feature_df.shape)

print()

display(feature_df.head())

print("\nSummary Statistics")
print("="*60)
print("Missing Values")
print("="*60)

print(feature_df.isnull().sum())
```
